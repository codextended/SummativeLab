```python
# 3) Aircraft category: ~64% of rows are missing a category.
#    Imputation: a manufacturer almost always builds a single type of aircraft
#    (Cessna -> Airplane, Robinson -> Helicopter, ...), so we fill missing values
#    with the most common known category for that Make. Make names are tidied
#    to title case first so that "CESSNA" and "Cessna" share the same lookup.
make_key = df_40_years["Make"].str.strip().str.title()
make_category_mode = (
    df_40_years.assign(make_key=make_key)
      .dropna(subset=["Aircraft.Category"])
      .groupby("make_key")["Aircraft.Category"]
      .agg(lambda s: s.mode()[0])
)
df_40_years["Aircraft.Category"] = df_40_years["Aircraft.Category"].fillna(make_key.map(make_category_mode))
```

**Construct metric for fatal/serious injuries**

*Hint:* Estimate the total number of passengers on each flight. The likelihood of serious / fatal injury can be estimated as a fraction from this.

```python
injury_cols = ["Total.Fatal.Injuries", "Total.Serious.Injuries",
               "Total.Minor.Injuries", "Total.Uninjured"]
df_40_years[injury_cols]
```

```python
## Missing Value in a category must be recorded as 0
df_40_years[injury_cols] = df_40_years[injury_cols].fillna(0)

## Estimating the number of person on board
## total_person = fatal_injuried + serious_injuried + minor_injuried + uninjured
df_40_years['Total.Passengers'] = df_40_years[injury_cols].sum(axis=1)
```

```python
## Let's count how many event has no passenger
(df_40_years["Total.Passengers"] == 0).sum()
```

```python
## Event with no passenger should not be keep
df_40_years = df_40_years[df_40_years['Total.Passengers'] > 0]
```

```python
## Likelihood of serious/fatal injuries
df_40_years['Fraction.Serious.Fatal'] = (df_40_years["Total.Fatal.Injuries"] + df_40_years["Total.Serious.Injuries"]) / df_40_years["Total.Passengers"]
```

### Investigate the *Make* column
- Identify cleaning tasks here

```python
df_40_years['Make']
```

**Make cleaning tasks**

Inspecting `Make` shows the same manufacturer spelled in many ways:
1. **Inconsistent case**: `CESSNA` vs `Cessna`, `PIPER` vs `Piper`. Fix by stripping whitespace and converting to title case.
2. **Corporate suffixes**: `Cessna Aircraft Co`, `Piper Aircraft Inc`, `Cirrus Design Corp.`, `Air Tractor Inc`. Fix by removing words like *Aircraft, Corp, Corporation, Inc, Co, Company, Industries, Design, Ltd*, and the punctuation around them.
4. **Rare makes**: many makes have only a handful of accidents, which is too few to compare. Keep only makes with at least **50** events.

```python
## 1- Let's regularize the `letters case`
df_40_years['Make'] = df_40_years['Make'].str.title().str.strip()
df_40_years['Make']
```

```python
## 2- Let us remove some suffixes from the makes
suffix_pattern = (r"\b(Aircraft|Aicraft|Airplane|Acft|Aviation|Avn|Corporation|Corp|Inc|Co|"
                  r"Company|Industries|Industrie|Industry|Ind|Design|Ltd|Gmbh|Llc|S\.?A\.?|Sa)\b\.?")

df_40_years['Make'] = df_40_years['Make'].str.replace(suffix_pattern, "", regex=True).str.strip()
df_40_years['Make']
```

```python
## 3- Keep Makes with a reasonable number (>=50)
makes = df_40_years['Make'].value_counts()
makes_to_keep = makes[makes >= 50].index
makes_to_keep
df_40_years = df_40_years[df_40_years['Make'].isin(makes_to_keep)]
df_40_years['Make'].value_counts()
```

### Cleaning other columns
- there are other columns containing data that might be related to the outcome of an accident. We list a few here:
- Engine.Type
- Weather.Condition

- Broad.phase.of.flight

Inspect and identify potential cleaning tasks in each of the above columns. Execute those cleaning tasks. 

**Note**: You do not necessarily need to impute or drop NaNs here.

```python
other_cols = ["Engine.Type", "Weather.Condition", "Number.of.Engines",
              "Purpose.of.flight", "Broad.phase.of.flight"]
df_40_years.head()
```

```python
## Engine Type
df_40_years['Engine.Type'].value_counts(dropna=False)
```

```python
## "UNK" and "Unk" are the same thing -> "Unknown"
df_40_years["Engine.Type"] = df_40_years["Engine.Type"].replace(
    {"UNK": "Unknown"})
df_40_years['Engine.Type'].value_counts(dropna=False)
```

```python
## Weather Condition
df_40_years['Weather.Condition'].value_counts(dropna=False)
```

```python
df_40_years["Weather.Condition"] = df_40_years["Weather.Condition"].replace({"UNK": "Unknown", "Unk": "Unknown"})
df_40_years['Weather.Condition'].value_counts(dropna=False)
```

```python
## Number of Engines
df_40_years['Number.of.Engines'].value_counts(dropna=False)
```

```python
# Number.of.Engines: an airplane can't have 0 engines - treat 0 as missing
df_40_years["Number.of.Engines"] = df_40_years["Number.of.Engines"].replace(0, np.nan)
df_40_years['Number.of.Engines'].value_counts(dropna=False)
```

```python
# Purpose of FLight
df_40_years['Purpose.of.flight'].value_counts(dropna=False)
```

```python
## Let's merge spelling variants of the same purpose
df_40_years["Purpose.of.flight"] = df_40_years["Purpose.of.flight"].replace({
    "Public Aircraft - Federal": "Public Aircraft",
    "Public Aircraft - State": "Public Aircraft",
    "Public Aircraft - Local": "Public Aircraft",
    "PUBS": "Public Aircraft",
    "Air Race/show": "Air Race show",
    "ASHO": "Air Race show",
})

## Let's group very rare purposes (< 50 events) into "Other" so groups have enough samples
purpose_counts = df_40_years["Purpose.of.flight"].value_counts()
rare_purposes = purpose_counts[purpose_counts < 50].index
df_40_years["Purpose.of.flight"] = df_40_years["Purpose.of.flight"].replace(dict.fromkeys(rare_purposes, "Other"))

df_40_years['Purpose.of.flight'].value_counts(dropna=False)
```

```python
## Broad phase of flight
df_40_years['Broad.phase.of.flight'].value_counts(dropna=False)
```

```python
## Let's merge "Other" and "Unknown"
df_40_years["Broad.phase.of.flight"] = df_40_years["Broad.phase.of.flight"].replace({"Other": "Unknown"})
df_40_years['Broad.phase.of.flight'].value_counts(dropna=False)
```
